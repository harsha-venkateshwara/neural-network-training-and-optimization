```python
import os
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Coverting to grey scale
#Converting to Tensors
#Normalise
from torchvision import transforms

transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=1), 
    transforms.ToTensor(),           
    transforms.Normalize((0.5,), (0.5,))
])
```

```python
#Creating a Pytorch dataset of images
dataset_path = "/Users/sourabhraja/Downloads/datasets/cnn_dataset"

dataset = datasets.ImageFolder(root=dataset_path, transform=transform)
```

```python
#Printing the number samples
print("Total samples:", len(dataset))             
print("Number of classes:", len(dataset.classes)) 
print("Class names:", dataset.classes)
```

```python
#Displaying subset of images
def show_batch(dataloader):
    images, labels = next(iter(dataloader))
    fig, axes = plt.subplots(3, 3, figsize=(6, 6))
    for i, ax in enumerate(axes.flatten()):
        img = images[i]
        img = img * 0.5 + 0.5
        
       
        img = img.permute(1, 2, 0)

        ax.imshow(img.squeeze(), cmap='gray' if img.shape[2] == 1 else None)
        ax.set_title(dataset.classes[labels[i]])
        ax.axis('off')
    plt.show()

show_batch(dataloader)
```

```python
#splitting the dataset into training (80%), validation (10%), and test (10%) sets 

from torch.utils.data import random_split, DataLoader


total_size = len(dataset)
train_size = int(0.8 * total_size)
temp_size = total_size - train_size
val_size = int(0.5 * temp_size)
test_size = temp_size - val_size

train_dataset, val_dataset, test_dataset = random_split(dataset, [train_size, val_size, test_size])
```

```python
#DataLoader objects created  for the training, validation, and test datasets, each loading data in batches of size 64
batch_size = 64

train_data_loaded = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
validation_data_loaded = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_data_loaded = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
#Defining the CNN module
class CNN(nn.Module):
    def __init__(self, in_channels=1, num_classes=36, dropout_p=0.4):
        super(CNN, self).__init__()
        
       
        self.conv1 = nn.Conv2d(in_channels, 16, kernel_size=3, padding=1)  
        self.conv2 = nn.Conv2d(16, 48, kernel_size=3, padding=1)           
        self.conv3 = nn.Conv2d(48, 96, kernel_size=3, padding=1)           
        self.pool = nn.MaxPool2d(2, 2)
        self.pool2 = nn.MaxPool2d(2, 2)

      
        self.fc1 = nn.Linear(96*7*7, 384)    
        self.dropout = nn.Dropout(dropout_p) 
        self.fc2 = nn.Linear(384, num_classes)

    def forward(self, x):
        x = F.leaky_relu(self.conv1(x))  
        x = F.leaky_relu(self.conv2(x))
        x = self.pool(x)
        x = F.leaky_relu(self.conv3(x))
        x = self.pool2(x)
        x = x.view(x.size(0), -1)         
        x = F.leaky_relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x) 
                                                  
        return x
                                
```

```python
#Q-3,4,5 Training loop
# The loop runs for 10 epochs as defined by the num_epochs variable.
# Each epoch consists of one full training pass (model.train()) followed by one full validation pass (model.eval()).
# During validation, gradient calculation is disabled using torch.no_grad() to save memory and computations.
# Training and validation loss/accuracy are calculated and printed at the end of each epoch.

# The total training time is measured and printed after the loop completes
import torch
import torch.nn as nn
import torch.optim as optim
import time

model = CNN(in_channels=1, num_classes=36, dropout_p=0.3)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

num_epochs = 10

start_time = time.time()

for epoch in range(num_epochs):

  
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for inputs, labels in train_loader:
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * inputs.size(0)
        _, preds = torch.max(outputs, 1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)

    trainingLoss = running_loss / total
    accuracy_epoch = correct / total


    model.eval()
    val_running_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for inputs, labels in val_loader:
            outputs = model(inputs)
            loss = criterion(outputs, labels)

            val_running_loss += loss.item() * inputs.size(0)
            _, preds = torch.max(outputs, 1)
            val_correct += (preds == labels).sum().item()
            val_total += labels.size(0)

    val_loss_epoch = val_running_loss / val_total
    val_acc_epoch = val_correct / val_total


    print(f"Epoch [{epoch+1}] - Training Loss: {trainingLoss:.4f}, "
          f"Train Accuracy: {accuracy_epoch:.4f}, Validation Loss: {val_loss_epoch:.4f}, "
          f"Validation Accuracy: {val_acc_epoch:.4f}")

end_time = time.time()
print(f"\nTraining model completed in: {end_time - start_time:.2f} s")


```

```python
#Improved model

import torch.nn.functional as F

class CNN(nn.Module):
    def __init__(self, in_channels=1, num_classes=36, dropout_p=0.5):
        super(CNN, self).__init__()
        
        self.conv1 = nn.Conv2d(in_channels, 16, kernel_size=3, padding=1)
        self.conv2 = nn.Conv2d(16, 48, kernel_size=3, padding=1)
        self.conv3 = nn.Conv2d(48, 96, kernel_size=3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.pool2 = nn.MaxPool2d(2, 2)
        
        
        self.dropout_conv = nn.Dropout2d(0.2)   
        self.dropout_fc = nn.Dropout(dropout_p) 

        self.fc1 = nn.Linear(96*7*7, 384)
        self.fc2 = nn.Linear(384, num_classes)

    def forward(self, x):
        x = F.leaky_relu(self.conv1(x))
        x = self.dropout_conv(x)
        x = F.leaky_relu(self.conv2(x))
        x = self.pool(x)
        x = self.dropout_conv(x)
        x = F.leaky_relu(self.conv3(x))
        x = self.pool2(x)
        x = x.view(x.size(0), -1)
        x = F.leaky_relu(self.fc1(x))
        x = self.dropout_fc(x)
        x = self.fc2(x)
        return x
```

```python
import time
import pickle
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torchvision import transforms

def train_with_methods(model_class, train_dataset, val_loader, methods,num_epochs=30, model_name="cnn_best", early_stopping_limit=5,batch_size=64):
   

    criterion = nn.CrossEntropyLoss()
    best_val_loss_overall = None
    best_model = None
    best_history = None
    best_method = None

    for i, method in enumerate(methods):
        print(f"Training with method {i+1}: {method} \n")

        
        if method.get("augment", False):
            transform_aug = transforms.Compose([
                transforms.Grayscale(num_output_channels=1),
                transforms.RandomRotation(15),
                transforms.RandomHorizontalFlip(),
                transforms.ToTensor(),
                transforms.Normalize((0.5,), (0.5,))
            ])
        else:
            transform_aug = transforms.Compose([
                transforms.Grayscale(num_output_channels=1),
                transforms.ToTensor(),
                transforms.Normalize((0.5,), (0.5,))
            ])

        
        train_dataset_aug = train_dataset
        train_dataset_aug.transform = transform_aug
        train_loader = torch.utils.data.DataLoader(train_dataset_aug, batch_size=batch_size, shuffle=True)

        
        model = model_class(**method.get("model_params", {}))
        optimizer = optim.Adam(model.parameters(), lr=method.get("lr", 0.001))
        early_stopping_counter = 0

        history = {
            "train_loss": [],
            "val_loss": [],
            "train_acc": [],
            "val_acc": []
        }

        start_time = time.time()

        for epoch in range(num_epochs):
            model.train()
            running_loss = 0.0
            correct = 0
            total = 0

            for inputs, labels in train_loader:
                optimizer.zero_grad()
                outputs = model(inputs)
                loss = criterion(outputs, labels)
                loss.backward()
                optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                _, preds = torch.max(outputs, 1)
                correct += (preds == labels).sum().item()
                total += labels.size(0)

            train_loss_epoch = running_loss / total
            train_acc_epoch = correct / total

           
            model.eval()
            val_loss = 0.0
            val_correct = 0
            val_total = 0
            with torch.no_grad():
                for val_inputs, val_labels in val_loader:
                    val_outputs = model(val_inputs)
                    loss = criterion(val_outputs, val_labels)
                    val_loss += loss.item() * val_inputs.size(0)
                    _, val_preds = torch.max(val_outputs, 1)
                    val_correct += (val_preds == val_labels).sum().item()
                    val_total += val_labels.size(0)

            val_loss_epoch = val_loss / val_total
            val_acc_epoch = val_correct / val_total

            history["train_loss"].append(train_loss_epoch)
            history["train_acc"].append(train_acc_epoch)
            history["val_loss"].append(val_loss_epoch)
            history["val_acc"].append(val_acc_epoch)

            
            if best_val_loss_overall is None or val_loss_epoch < best_val_loss_overall:
                best_val_loss_overall = val_loss_epoch
                best_model = model
                best_history = history
                best_method = method
                torch.save(model.state_dict(), f"{model_name}_best.pth")
                with open(f"{model_name}_weights.pkl", "wb") as f:
                    pickle.dump(model.state_dict(), f)
                early_stopping_counter = 0
            else:
                early_stopping_counter += 1

            print(f"Epoch [{epoch+1}] - Train Loss: {train_loss_epoch:.4f}, Train Acc: {train_acc_epoch:.4f}, "
                  f"Val Loss: {val_loss_epoch:.4f}, Val Acc: {val_acc_epoch:.4f}")

            if early_stopping_counter >= early_stopping_limit:
                print(f"Early stopping triggered for method {i+1} at epoch {epoch+1}.")
                break

        print(f"Method {i+1} completed in {time.time() - start_time:.2f}s")

    print(f"\nBest method: {best_method}")
    return best_model, best_history, best_method
```

```python
dataset_path = "/Users/sourabhraja/Downloads/datasets/cnn_dataset"

dataset = datasets.ImageFolder(root=dataset_path)
```

```python
# Loops through a list of different training methods i.e varying augmentations, learning rates, or model parameters to find the best configuration.
#For each method, it tracks detailed training and validation loss/accuracy in a history dictionary and early stopping  is used to stop if validation performance doesn't improve.

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import random_split, DataLoader, Dataset
from torchvision import transforms
from PIL import Image

class TransformSubset(Dataset):
    def __init__(self, subset, transform=None):
        self.subset = subset
        self.transform = transform
    
    def __getitem__(self, index):
        x, y = self.subset[index]
        if self.transform:
            x = self.transform(x)
        return x, y
    
    def __len__(self):
        return len(self.subset)
        
basic_transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),  
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])


aug_transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=1), 
    transforms.RandomRotation(15),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])



total_size = len(dataset)
train_size = int(0.8 * total_size)
val_size = int(0.1 * total_size)
test_size = total_size - train_size - val_size

train_subset, val_subset, test_subset = random_split(dataset, [train_size, val_size, test_size])


train_dataset = TransformSubset(train_subset, aug_transform)
val_dataset = TransformSubset(val_subset, basic_transform)
test_dataset = TransformSubset(test_subset, basic_transform)  


```

```python
batch_size = 64
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
'''All three methods will share these settings:

Max Epochs: Each method will be trained for a maximum of 50 epochs.

Early Stopping: For all three methods, training will automatically stop if the validation loss does not improve for 5 consecutive epochs.

Model Saving: The function will compare the final validation loss from all three methods and save the single best model as cnn_augmented_best.pth and cnn_augmented_weights.pkl.

Breakdown:

Method 1: Dropout at 0.4

Model: A CNN is created with a dropout rate of 0.4.

Augmentation: False. The model will only see the original, unchanged training images.

Learning Rate: 0.001

Method 2: Dropout at 0.5

Model: A new CNN is created with a higher dropout rate of 0.5.

Augmentation: False. This method also trains without data augmentation.

Learning Rate: 0.001

Purpose: This directly compares how increasing dropout from 0.4 to 0.5 affects performance, keeping everything else the same.

Method 3: Dropout at 0.5 + Augmentation

Model: A third CNN is created, also with a dropout rate of 0.5.

Augmentation: True. This is the key difference. This model will be trained on images that are randomly rotated and flipped, helping it generalize better.

Learning Rate: 0.001

Purpose: This tests the combined effect of high dropout (0.5) and data augmentation.'''
```

```python


methods = [
    {"model_params": {"in_channels":1, "num_classes":36, "dropout_p":0.4}, "lr":0.001, "augment": False}, 
    {"model_params": {"in_channels":1, "num_classes":36, "dropout_p":0.5}, "lr":0.001, "augment": False}, 
    {"model_params": {"in_channels":1, "num_classes":36, "dropout_p":0.5}, "lr":0.001, "augment": True}   
]
best_model, best_history, best_method = train_with_methods(
    CNN,
    train_dataset,
    val_loader,
    methods,
    num_epochs=50,
    model_name="cnn_augmented",
    early_stopping_limit=5,
    batch_size=batch_size
)
```

```python
# Testing the model with the best model obtained from the 3 different parameter methods.
import torch
import numpy as np
from sklearn.metrics import precision_recall_fscore_support
import time


model = CNN(in_channels=1, num_classes=36, dropout_p=0.5)
model.load_state_dict(torch.load("best_model.pth"))
model.eval()


y_true = []
y_predictions = []
test_loss = 0.0
test_correct = 0
test_total = 0

criterion = nn.CrossEntropyLoss()

start_time = time.time()

with torch.no_grad():
    for inputs, labels in test_loader:
        inputs, labels = inputs, labels
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        test_loss = test_loss + loss.item() * inputs.size(0)
        _, preds = torch.max(outputs, 1)
        y_predictions.extend(preds.cpu().numpy())
        test_correct += (preds == labels).sum().item()
        y_true.extend(labels.numpy())
        test_total = test_total + labels.size(0)

test_time = time.time() - start_time
test_accuracy = test_correct / test_total
test_loss = test_loss / test_total

y_true = np.array(y_true)
y_predictions = np.array(y_predictions)

precision, recall, f1_score, support = precision_recall_fscore_support(
    y_true, y_predictions, average='weighted', zero_division=0
)

print(f"\nTest Time: {test_time:.2f} seconds")
print(f"\nTest Data Accuracy: {test_accuracy * 100:.2f}%")
print(f"\nTest Data Loss: {test_loss:.4f}")
print(f"\nPrecision: {precision:.4f}")
print(f"\nRecall: {recall:.4f}")
print(f"\nF1 Score: {f1_score:.4f}")
```
