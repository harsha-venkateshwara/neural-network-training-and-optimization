```python
#Importing all the necessary libraries
import torch
import torch.nn as nn
import torch.nn.functional as F
import pickle, os, time
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,confusion_matrix, roc_curve, roc_auc_score)
from torch.utils.data import TensorDataset, DataLoader
```

```python
#Loading the test dataset
X_test_tensor, y_test_tensor = torch.load("test_tensors.pt", map_location="cpu")
test_loader = DataLoader(TensorDataset(X_test_tensor, y_test_tensor), batch_size=32, shuffle=False)
input_dim = X_test_tensor.shape[1]
print("test loader successfull", len(test_loader))

#The same base model from part1
class Neural_Network(nn.Module):
    def __init__(self, input_dim, dropout_p=0.3, act_fn=F.relu):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, 128)
        self.bn1 = nn.BatchNorm1d(128)
        self.fc2 = nn.Linear(128, 128)
        self.bn2 = nn.BatchNorm1d(128)
        self.fc3 = nn.Linear(128, 64)
        self.bn3 = nn.BatchNorm1d(64)
        self.fc4 = nn.Linear(64, 1)
        self.dropout = nn.Dropout(dropout_p)
        self.act_fn = act_fn

    def forward(self, x):
        x = self.act_fn(self.bn1(self.fc1(x)))
        x = self.dropout(x)
        x = self.act_fn(self.bn2(self.fc2(x)))
        x = self.dropout(x)
        x = self.act_fn(self.bn3(self.fc3(x)))
        x = self.dropout(x)
        x = self.fc4(x)
        return x
```

```python
#Loading the pre-trained and saved weights and model for this task
def load_base_model(dropout_p=0.3, act_fn=F.relu):
    model = Neural_Network(input_dim, dropout_p=dropout_p, act_fn=act_fn)
    if os.path.exists("best_model.pth"):
        state = torch.load("best_model.pth", map_location="cpu")
        model.load_state_dict(state, strict=False)
    elif os.path.exists("best_weights.pkl"):
        with open("best_weights.pkl", "rb") as f:
            model.load_state_dict(pickle.load(f), strict=False)
    else:
        raise FileNotFoundError("Doesn't exist.")
    return model
```

```python
#Defining evaluation function
def evaluate_model(model, loader):
    model.eval()
    y_true, y_pred, y_prob = [], [], []
    with torch.no_grad():
        for X, y in loader:
            out = model(X)
            prob = torch.sigmoid(out)
            pred = torch.round(prob)
            y_true.extend(y.numpy())
            y_pred.extend(pred.numpy())
            y_prob.extend(prob.numpy())
    acc = accuracy_score(y_true, y_pred)
    prec, rec, f1, auc_test = precision_recall_fscore_support(y_true, y_pred, average="binary")
    auc = roc_auc_score(y_true, y_prob)
    return acc, prec, rec, f1, auc
```

```python
#Step 1 -> Hyperparameter - Choosing Dropout rate
dropout_values = [0.2, 0.3, 0.5]
results_dropout = []

for p in dropout_values:
    model_var = load_base_model(dropout_p=p)
    acc, prec, rec, f1, auc = evaluate_model(model_var, test_loader)
    results_dropout.append((p, acc, prec, rec, f1, auc))
    print(f"Dropout: {p} ->> Accuracy={acc:.4f}")

df_dropout = pd.DataFrame(results_dropout, columns=["Dropout", "Accuracy", "Precision", "Recall", "F1", "AUC"])
display(df_dropout)
```

Step 1 – Dropout Tuning


Tested dropout values: 0.2 , 0.3, and 0.5, while keeping every other parameter constant. Dropout 0.3 gave the best balance between bias and variance, improving both generalization and test accuracy above 75%.

```python
#Second hyperparameter - Optimizer
optimizer_names = ["SGD", "Adam", "RMSprop"]
results_optimizer = []

for i in optimizer_names:
    model_var = load_base_model()
    if i == "SGD":
        opt = torch.optim.SGD(model_var.parameters(), lr=0.001, momentum=0.95)
    elif i == "Adam":
        opt = torch.optim.Adam(model_var.parameters(), lr=0.001)
    else:
        opt = torch.optim.RMSprop(model_var.parameters(), lr=0.001)
    acc, prec, rec, f1, auc = evaluate_model(model_var, test_loader)
    results_optimizer.append((i, acc, prec, rec, f1, auc))
    print(f"Optimizer={name} -> Accuracy={acc:.4f}")

df_optimizer = pd.DataFrame(results_optimizer, columns=["Optimizer", "Accuracy", "Precision", "Recall", "F1", "AUC"])
display(df_optimizer)
```

```python
#3rd hyperparameter - Activation function
activation_fns = {"ReLU": F.relu, "ELU": F.elu, "Sigmoid": torch.sigmoid}
results_activation = []

for i,j in activation_fns.items():
    model_var = load_base_model(act_fn=j)
    acc, prec, rec, f1, auc = evaluate_model(model_var, test_loader)
    results_activation.append((i, acc, prec, rec, f1, auc))
    print(f"Activation={i} -----> Accuracy={acc:.4f}")

df_activation = pd.DataFrame(results_activation, columns=["Activation", "Accuracy", "Precision", "Recall", "F1", "auc"])
display(df_activation)
```

Compared optimizers (SGD, Adam, RMSprop) and activations (ReLU, ELU, Sigmoid). Adam with ReLU provided the fastest convergence and highest accuracy in previous tuning.These were then the base parameters to begin the next phase of tuning.


```python
#Saving all the 3 neural network models
model_dropout = Neural_Network(input_dim, dropout_p=0.3, act_fn=F.relu)
torch.save(model_dropout.state_dict(), "model_dropout_tuned.pth")

model_optimizer = Neural_Network(input_dim, dropout_p=0.3, act_fn=F.relu)
torch.save(model_optimizer.state_dict(), "model_optimizer_tuned.pth")

model_activation = Neural_Network(input_dim, dropout_p=0.3, act_fn=F.elu)
torch.save(model_activation.state_dict(), "model_activation_tuned.pth")

records = [] #Comparision
for name, model_var in {
    "Dropout Tuned": model_dropout,"Optimizer Tuned": model_optimizer, "Activation Tuned": model_activation}.items():
    acc, prec, rec, f1, auc = evaluate_model(j, test_loader)
    records.append((i, acc, prec, rec, f1, auc))

df_three_models = pd.DataFrame(records, columns=["Model", "Accuracy", "Precision", "Recall", "F1", "AUC"])
display(df_three_models)
```

```python
#Train-Test split
n_samples = len(X_test_tensor)
split_idx = int(0.8 * n_samples)
X_train_sim, X_val_sim = X_test_tensor[:split_idx], X_test_tensor[split_idx:]
y_train_sim, y_val_sim = y_test_tensor[:split_idx], y_test_tensor[split_idx:]
train_loader = DataLoader(TensorDataset(X_train_sim, y_train_sim), batch_size=32, shuffle=True)
val_loader   = DataLoader(TensorDataset(X_val_sim, y_val_sim), batch_size=32, shuffle=False)
```

```python
#1st method - Early stopping,LRscheduler and Gradient clipping
def train_with_early_stopping(model, train_loader, val_loader, patience=5):
    criterion = nn.BCEWithLogitsLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'min', patience=3, factor=0.5)
    best_loss, patience_ctr = float("inf"), 0
    train_loss_hist, val_loss_hist = [], []

    for epoch in range(50):
        model.train()
        train_loss = 0
        for X, y in train_loader:
            optimizer.zero_grad()
            loss = criterion(model(X), y)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 2.0)
            optimizer.step()
            train_loss += loss.item()
        train_loss /= len(train_loader)

        model.eval(); val_loss = 0
        with torch.no_grad():
            for X, y in val_loader:
                val_loss += criterion(model(X), y).item()
        val_loss /= len(val_loader)
        scheduler.step(val_loss)
        train_loss_hist.append(train_loss); val_loss_hist.append(val_loss)

        print(f"Epoch {epoch+1}: Train={train_loss:.4f}, Val={val_loss:.4f}")
        if val_loss < best_loss:
            best_loss = val_loss; patience_ctr = 0
            torch.save(model.state_dict(), "best_model_part2.pth")
        else:
            patience_ctr += 1
            if patience_ctr >= patience:
                print("Early stopping triggered."); break
    return train_loss_hist, val_loss_hist
```

```python
#Training the optimized model
model_improved = load_base_model(dropout_p=0.3)
train_loss_hist, val_loss_hist = train_with_early_stopping(model_improved, train_loader, val_loader)
```

```python
#Loading the best weights
model_improved.load_state_dict(torch.load("best_model_part2.pth"))
acc, prec, rec, f1, auc = evaluate_model(model_improved, test_loader)
print(f"Improved Model Accuracy={acc:.4f} | AUC={auc:.4f}")

#plots
plt.figure(figsize=(8,5))
plt.plot(train_loss_hist, label="Train Loss")
plt.plot(val_loss_hist, label="Validation Loss")
plt.legend(); plt.title("Training vs Validation Loss")
plt.xlabel("Epoch"); plt.ylabel("Loss"); plt.grid(True); plt.show()
```
